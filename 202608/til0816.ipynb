- `describe()` 메서드: 데이터프레임에 대한 기술통계를 자동으로 추출해 주는 메서드  
- `count`: 누락된 값을 제외한 데이터 개수  
- `mean`: 평균  
- `std`: 표준편차  
- `min`: 최솟값
- `max`: 최댓값  
- 0.25, 0.5, 0.75: 데이터를 순서대로 늘어 놓았을 때 25%, 50%, 75% 지점에 놓인 값

```python
ns_book6.describe()
```

```python
# 도서권수가 0인 행 제외
ns_book7 = ns_book6[ns_book6['도서권수'] > 0]
```

- `describe()` 메서드의 `percentiles` 매개변수에 위치를 지정하면 원하는 위치에 있는 값을 볼 수 있음

```python
ns_book7.describe(percentiles=[0.3, 0.6, 0.9])
```

- `describe()` 메서드의 `include` 매개변수에 데이터 타입 지정 가능

```python
ns_book7.describe(include='object') # top: 가장 많이 등장하는 항목, freq: 가장 많이 등장하는 항목의 빈도수
```

- `mean()` 메서드: 평균을 계산하는 메서드  
- `median()` 메서드: 중앙값을 계산하는 메서드  
    - 데이터 개수가 짝수일 때는 가운데 두 값을 평균하여 중앙값을 결정  
- `max()` 메서드: 최댓값을 계산하는 메서드  
- `min()` 메서드: 최솟값을 계산하는 메서드  
- `quantile()` 메서드: 원하는 분위수 값을 계산하는 메서드  
- `var()` 메서드: 분산을 계산하는 메서드  
- `std()` 메서드: 표준편차를 계산하는 메서드  
- `mode()` 메서드: 최빈값을 계산하는 메서드  

```python
ns_book7['대출건수'].mean()
```

```python
ns_book7['대출건수'].median()
```

```python
ns_book7['대출건수'].max()
```

```python
ns_book7['대출건수'].quantile(0.25)
```

- `quantile()` 메서드의 `interpolation` 매개변수를 지정하면 `quantile`을 계산하는 방식을 지정 가능  
    - `linear`(default): 양쪽 분위수에 비례하여 결정  
    - `midpoint`: 분위수에 상관없이 무조건 두 수 사이의 중앙값 사용  
    - `nearest`: 두 수 중에서 가까운 값을 선택  

```python
print(pd.Series([1,2,3,4,5]).quantile(0.9))
print(pd.Series([1,2,3,4,5]).quantile(0.9, interpolation='midpoint'))
print(pd.Series([1,2,3,4,5]).quantile(0.9, interpolation='nearest'))
```

- 기술통계를 구하는 메서드에서 `numeric_only` 매개변수를 `True`로 지정하면 수치형 열만 연산 가능  
- `True`로 지정하지 않으면 모든 데이터 타입의 열에 대해 수행하기 때문에 시간이 오래 걸림

```python
ns_book7.mean()
```

```python
ns_book7.mean(numeric_only=True)
```

- `numpy` 라이브러리를 이용해도 여러 기술통계를 구할 수 있음  

```python
import numpy as np

# 평균
print(np.mean(ns_book7['대출건수']))   

# 가중 평균
print(np.average(ns_book7['대출건수'], weights=1/ns_book7['도서권수']))    # weights 매개변수로 가중치 지정

# 중앙값
print(np.median(ns_book7['대출건수']))

# 최댓값
print(np.max(ns_book7['대출건수']))

# 분위수
print(np.quantile(ns_book7['대출건수'], [0.25,0.5,0.75]))
print(np.percentile(ns_book7['대출건수'], [75]))

# 분산
print(ns_book7['대출건수'].var())   # 판다스는 디폴트 자유도가 n-1
print(np.var(ns_book7['대출건수'])) # 넘파이는 자유도를 n으로 계산
print(np.var(ns_book7['대출건수'], ddof=1)) # 자유도를 n으로 계산

# 표준편차
print(np.std(ns_book7['대출건수']))

# 최빈값
values, counts = np.unique(ns_book7['도서명'], return_counts=True)  # 별도의 함수를 제공하지 않으므로 unique()를 이용해 고유값과 등장 횟수를 얻음
max_idx = np.argmax(counts) # argmax()를 이용하여 가장 많은 값의 인덱스를 탐색
print(values[max_idx])
```
