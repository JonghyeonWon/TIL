````python
import gdown
gdown.download('https://bit.ly/3pK7iuu', 'ns_book7.csv', quiet=False)
````

````python
import pandas as pd
ns_book7 = pd.read_csv('ns_book7.csv', low_memory=False)
````

````python
import matplotlib.pyplot as plt
````

- `Figure` 객체: 모든 그래프 구성 요소를 담고 있는 최상위 객체
- `figure()` 함수로 피겨 객체를 만들어 활용하면 다양한 그래프 옵션 조절 가능

````python
plt.scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
plt.show()
````

````python
plt.figure(figsize=(9, 6))  # 너비가 9인치, 높이가 6인치인 figure 객체 생성
plt.scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
plt.show()
````

- 타이트 레이아웃: 그래프를 출력할 때 그래프 주변 공백 최소화  
- `bbox_inches` 옵션: 타이트 레이아웃 사용 여부(`None`이면 사용하지 않음)

```python
%config InlineBackend.print_figure_kwargs = {'bbox_inches' : None}
%config InlineBackend.print_figure_kwargs = {'bbox_inches' : 'tight'}
```

- `dpi(dot per inch)`: 1인치를 몇 개의 점(픽셀)으로 표현하는지를 나타냄  
- `figure()` 함수의 `dpi` 매개변수를 통해 DPI 지정 가능

````python
# dpi가 커지면 그래프 사이즈가 커짐
plt.figure(dpi=144)
plt.scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
plt.show()
````

- `rcParams` 객체: `matplotlib`의 기본값을 관리하는 객체  
- 이를 통해 변경하면 이후에 그려지는 모든 그래프에 바뀐 설정이 적용됨  

````python
# dpi 기본값 변경
plt.rcParams['figure.dpi'] = 100
````

````python
# 산점도 마커 모양 변경
plt.rcParams['scatter.marker']  # 현재 모양인 동그라미 표시
````

````python
plt.rcParams['scatter.marker'] = '*'
plt.scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
plt.show()
````

- `rcParams` 객체 변경 대신 `marker` 매개변수로 마커 모양을 지정하면 각각의 그래프에 대한 모양만 지정 가능

````python
plt.scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1, marker='+')
plt.show()
````

- 서브플롯(subplot): `matplotlib`의 `axes` 클래스의 객체를 의미  
- 하나의 서브플롯은 두 개 이상의 축 포함  
- 각 축에는 눈금(tick)과 레이블(label) 존재  

- 서브플롯을 정의하려면 `subplots()` 함수에 원하는 플롯 개수를 지정

````python
fig, axs = plt.subplots(2)
axs[0].scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
axs[1].hist(ns_book7['대출건수'], bins=100)
axs[1].set_yscale('log')

fig.show()
````

- `figsize` 매개변수: `subplots()` 함수 내에서 figure`의 크기를 지정할 수 있음  
- `set_title()` 메서드: 각 그래프의 제목을 지정  

````python
fig, axs = plt.subplots(2, figsize=(6,8))
axs[0].scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
axs[0].set_title('scatter plot')

axs[1].hist(ns_book7['대출건수'], bins=100)
axs[1].set_title('histogram')
axs[1].set_yscale('log')

fig.show()
````

- `subplots()` 함수에 행과 열을 지정하면 원하는 서브플롯 개수의 figure를 만들 수 있음  
- 첫 번째 매개변수는 행의 개수, 두 번째 매개변수는 열의 개수  

````python
fig, axs = plt.subplots(1, 2, figsize=(10,4))
axs[0].scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
axs[0].set_title('scatter plot')
axs[0].set_xlabel('number of books')
axs[0].set_ylabel('borrow count')

axs[1].hist(ns_book7['대출건수'], bins=100)
axs[1].set_title('histogram')
axs[1].set_yscale('log')
axs[1].set_xlabel('borrow count')
axs[1].set_ylabel('frequency')

fig.show()
````
