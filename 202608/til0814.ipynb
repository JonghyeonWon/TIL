```python
# 남산도서관 데이터 다운로드
import gdown
gdown.download('https://bit.ly/3RhoNho', 'ns_202104.csv', quiet=False)  # quiet=False: 진행 사항을 표시
```

```python
import pandas as pd

ns_df = pd.read_csv('ns_202104.csv', low_memory=False)  # low_memory=False: 각 컬럼의 타입을 정확하게 추론하기 위한 옵션
ns_df.head()
```

```python
# 'Unnamed: 13' 컬럼 제외하기
ns_book = ns_df.loc[:, '번호':'등록일자']
ns_book.head()
```

```python
print(ns_df.columns)
print(ns_df.columns[0])
```

- Index 클래스를 비롯하여 판다스 배열 성격의 객체는 어떤 값과 비교할 때 자동으로 배열에 있는 모든 원소와 한 번씩 비교 -> **원소별 비교**

```python
print(ns_df.columns != 'Unnamed: 13')
print(type(ns_df.columns != 'Unnamed: 13'))
```

- 이렇게 생성된 객체의 결과는 넘파이 배열(Numpy array)

```python
# 비교 결과로 배열 생성
selected_columns = ns_df.columns != 'Unnamed: 13'
ns_book = ns_df.loc[:, selected_columns]    # 열의 조건으로 unnamed가 아닌 열만 추출
ns_book.head()
```

- `drop()` 메서드: 데이터프레임의 행이나 열을 삭제하는 메서드
    - 첫 번째 매개변수로는 삭제하려는 행/열의 이름, `axis` 매개변수가 0이면 행, 1이면 열을 의미

```python
ns_book = ns_df.drop(['부가기호', 'Unnamed: 13'], axis = 1)   
ns_book.head(2)
```

pd의 메서드의 매개변수 중 `inplace` 매개변수를 `True`로 지정하면 수정한 결과를 바로 데이터프레임에 반영 가능

```python
ns_book.drop('주제분류번호', axis=1, inplace=True)
ns_book.head(2)
```

- `dropna()` 메서드: `NaN`이 포함된 행 또는 열을 삭제

```python
ns_book = ns_df.dropna(axis = 1)    # NaN이 하나라도 포함된 열 삭제
ns_book.head(2)
```

```python
ns_book = ns_df.dropna(axis=1, how='all')   # 모든 값이 NaN인 열을 삭제하려면 how를 'all'로 지정
ns_book.head(2)
```

```python
ns_book2 = ns_book.drop([0,1])  # 0, 1번째 행 삭제
ns_book2.head(2)
```

- `duplicated()` 메서드: 중복된 행 중에서 처음 행을 제외한 나머지 행은 `True`,  
                        그 외에 중복되지 않은 나머지 모든 행은 `False`로 표시

```python
print(ns_book.duplicated())
print(sum(ns_book.duplicated()))   # 0 -> True인 값이 없음을 의미하므로 중복된 행이 없음
```

- 일부 열을 기준으로 중복된 행을 찾으려면 `subset` 매개변수에 기준 열을 나열

```python
sum(ns_book.duplicated(subset=['도서명','저자','ISBN']))
```

```python
dup_rows = ns_book.duplicated(subset=['도서명','저자','ISBN'], keep=False)  # keep=False 옵션은 중복되는 모든 행에 대해 True로 표시
ns_book3 = ns_book[dup_rows]
ns_book3.head(3)
```

```python
# groupby() 메서드 사용예시
count_df = ns_book[['도서명','저자','ISBN','권','대출건수']]

# by 매개변수를 통해 기준이 되는 열 지정
# dropna=False로 지정하면 NaN이 있는 행도 포함하여 계산
group_df = count_df.groupby(by=['도서명','저자','ISBN','권'], dropna=False)
loan_count = group_df.sum()
```

```python
loan_count = count_df.groupby(by=['도서명','저자','ISBN','권'], dropna=False).sum()
loan_count.head()
```

- `~` 연산자를 사용하면 중복된 행을 `True`로 표시한 `boolean` 배열을 반전 가능

```python
dup_rows = ns_book.duplicated(subset=['도서명','저자','ISBN','권'])

unique_rows = ~dup_rows 

# 중복되지 않는 행만 copy
ns_book3 = ns_book[unique_rows].copy()
```

```python
sum(ns_book3.duplicated(subset=['도서명','저자','ISBN','권']))  # sum의 결과가 0이므로 중복되는 행 없음
```

- `set_index()` 메서드: 지정한 열을 기준으로 인덱스를 설정하는 메서드

```python
ns_book3.set_index(['도서명','저자','ISBN','권'], inplace=True)
ns_book3.head()
```

- `update()` 메서드: 다른 데이터프레임을 이용해 원본 데이터프레임의 값을 업데이트하는 메서드

```python
ns_book3.update(loan_count)
ns_book3.head()
```

- `reset_index()` 메서드: 데이터프레임의 인덱스를 재설정하는 메서드 

```python
ns_book4 = ns_book3.reset_index()
ns_book4.head()
```

```python
ns_book4 = ns_book4[ns_book.columns]    # 원래 데이터의 컬럼과 일치하는 컬럼만 선택
ns_book4.head(3)
```

```python
ns_book4.to_csv('ns_book4.csv', index=False)    # csv파일로 저장
```
